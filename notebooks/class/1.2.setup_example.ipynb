## Lesson summary

This classroom setup notebook gives students a minimal repeatable check that the course kernel and core dependencies are available before moving to market data work.

## Learning objectives
- Install the classroom environment dependencies.
- Verify data, finance, and storage libraries such as `pandas`, `yfinance`, `fredapi`, and `pyarrow`.
- Create an environment record that can be attached to class submissions.

## Lesson flow
1. Clone the setup assets.
2. Install the requirements.
3. Validate the runtime with `watermark`.


## Setup

Open the notebook from JupyterLab after running `uv sync`; the runtime cells should execute without downloading market data.


## Terminal setup with pyenv and uv

Run these commands from the repository root before opening JupyterLab. The project dependencies are declared in `pyproject.toml` and installed by `uv sync`.

```bash
pyenv install 3.12.12
pyenv local 3.12.12
uv sync
uv run jupyter lab
```
