## Runtime check

Run the following cells inside the `uv` environment to confirm that the active kernel is using Python 3.12 or newer and that the notebook can see the repository root.



````python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
for candidate in (project_root, *project_root.parents):
    if (candidate / "pyproject.toml").exists() and (candidate / "uv.lock").exists():
        project_root = candidate
        break

print(sys.version)
assert sys.version_info >= (3, 12)
assert (project_root / "pyproject.toml").exists()
assert (project_root / "uv.lock").exists()

````
