## Lesson summary

This classroom setup notebook gives students a short repeatable check before a live session. It verifies that the course kernel and core dependencies are available before moving to market data, exploratory analysis, and dashboard notebooks.

## Learning objectives

By the end of this note, students should be able to:

- confirm that the classroom environment has been synchronized from `uv.lock`;
- verify data, finance, visualization, and storage libraries such as `pandas`, `yfinance`, `fredapi`, `plotly`, `ipywidgets`, and `pyarrow`;
- create a concise environment record that can accompany reproducibility questions.

## Lesson flow

1. Open the repository from the project environment.
2. Run the Python version assertion.
3. Record the package versions with `watermark`.
4. Move to data notebooks only after the check passes.



## Setup

Open the notebook from JupyterLab after running `uv sync`; the runtime cells should execute without downloading market data or printing credentials.



## Terminal setup with pyenv and uv

Run these commands from the repository root before opening JupyterLab. The project dependencies are declared in `pyproject.toml` and installed by `uv sync`.

```bash
pyenv install 3.12.12
pyenv local 3.12.12
uv sync
uv run jupyter lab
```


## Runtime and package check

Use this check before running data notebooks that depend on external providers, interactive widgets, or columnar storage libraries.



````python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
for candidate in (project_root, *project_root.parents):
    if (candidate / "pyproject.toml").exists() and (candidate / "uv.lock").exists():
        project_root = candidate
        break

print(sys.version)
assert sys.version_info >= (3, 12)
assert (project_root / "notebooks" / "class").exists()

````
